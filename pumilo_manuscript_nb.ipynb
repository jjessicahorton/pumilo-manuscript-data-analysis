```python
def figure_1(figure, gal4, uas, knockdown_overexpression):
    fig, ax = plt.subplots(figsize = (10, 10))
    figure_1 = db.load(figure, idx = (knockdown_overexpression, gal4, uas), resamples = 1000)
    fig_1_df = pd.DataFrame(figure)
    fig_1_st = figure_1.cliffs_delta.statistical_tests
    fig_1_cd = figure_1.cliffs_delta
    fig_1_cdp = figure_1.cliffs_delta.plot(raw_marker_size = 0.5, swarm_label = "Y Axis", ax = ax)
    figure_variable_df_name = [name for name, value in globals().items() if value is figure][0]
    fig.savefig(f"{figure_variable_df_name}.svg", bbox_inches = 'tight', transparent = True)
    return fig_1_df.head(), fig_1_cd, fig_1_cdp, fig_1_st
```

```python
# Cliff's Delta and Permutation T-Test for Figure 1A
cd_1a = figure_1(figure_1a, "GAL4 only", "UAS only", "Knockdown")
cd_1a
```

```python
# Cliffs Delta & Permutation T Test, Figure 1B
cd_1b = figure_1(figure_1b, "GAL4 only", "UAS only", "Knockdown")
cd_1b
```

```python
# Cliffs Delta & Permutation T Test, Figure 1C
cd_1c = figure_1(figure_1c, "GAL4 only", "UAS only", "Knockdown")
cd_1c
```

## Figures 4A & 4B: Hypersentiziation Assay

- 4A: *pum* knockdown (HMS01564)
- 4B: *pum* overexpression

```python
figure_4a = pd.read_csv(r"C:\Users\jjess\Documents\MEGA\Bellemer_Lab\data_analysis\sara\figure_4A_updated.csv")
figure_4b = pd.read_csv(r"C:\Users\jjess\Documents\MEGA\Bellemer_Lab\data_analysis\sara\figure_4B_new_updated.csv")
```

- Six groups should be plotted in order given in the sheet
- GAL4 (UV-) should be compared to GAL4 (UV+) and so one for UAS and Knockdown using Cliff's Delta and Permutation T-Test
- Dot plots and violin plots (Gardner-Altman Plots)


```python
def figure_4(figure, gal4_neg, gal4_pos, uas_neg, uas_pos, kd_oe_neg, kd_oe_pos):
    fig, ax = plt.subplots(figsize = (10, 10))
    figure_4 = db.load(figure, idx = ((gal4_neg, gal4_pos,), (uas_neg, uas_pos,), (kd_oe_neg, kd_oe_pos,)), resamples = 1000)
    fig_4_df = pd.DataFrame(figure)
    fig_4_st = figure_4.cliffs_delta.statistical_tests
    fig_4_cd = figure_4.cliffs_delta
    fig_4_cdp = figure_4.cliffs_delta.plot(raw_marker_size = 0.5, swarm_label = "Y Axis", custom_palette = "Paired", ax = ax)
    figure_variable_df_name = [name for name, value in globals().items() if value is figure][0]
    fig.savefig(f"{figure_variable_df_name}.svg", bbox_inches = 'tight', transparent = True)
    return fig_4_df.head(), fig_4_cd, fig_4_cdp, fig_4_st 
```

```python
# Cliff's Delta and Permutation T-Test for Figure 4A
cd_4a = figure_4(figure_4a, "GAL4 only (UV-)", "GAL4 only (UV+)", "UAS only (UV-)", "UAS only (UV+)", "Knockdown (UV-)", "Knockdown (UV+)")
cd_4a
```

```python
# Cliff's Delta and Permutation T-Test for Figure 4A
cd_4b = figure_4(figure_4b, "GAL4 only (UV-)", "GAL4 only (UV+)", "UAS only (UV-)", "UAS only (UV+)", "Overexpression (UV-)", "Overexpression (UV+)")
cd_4b
```
